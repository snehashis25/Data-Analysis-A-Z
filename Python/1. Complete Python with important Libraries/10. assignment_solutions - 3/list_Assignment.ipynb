```python
## 9
def transpose_matrix(matrix):
    return [[matrix[j][i] for j in range(len(matrix))] for i in range(len(matrix[0]))]


matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
transposed = transpose_matrix(matrix)
print("Original matrix:")
for row in matrix:
    print(row)
print("Transposed matrix:")
for row in transposed:
    print(row)
```

```python
## 10
def flattern_list(nested_list):
    return [item for sublist in nested_list for item in sublist]

nested_list = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
flattered_list = flattern_list(nested_list)
print(f"Nested list: {nested_list}")
print(f"Flatterned list: {flattered_list}")
```

```python
## 13
def reversed_list(lst):
    return lst[::-1]
lst = [1,2,3,4,5]
reversed_lst = reversed_list(lst)
print(f"Original list: {lst}")
print(f"Reversed list: {reversed_lst}")
```

```python
## 14
def rotated_list(lst, n):
    return lst[n:] + lst[:n]
lst = [1,2,3,4,5]
n = 2
rotated_list = rotated_list(lst, n)
print(f"Original List: {lst}")
print(f"Rotated List: {rotated_list}")
```

```python
## 15
def common_elements(list1, list2):
    return list(set(list1) & set(list2))
list1 = [1,2,3,4,5]
list2 = [4,5,6,7,8]
common = common_elements(list1, list2)
print(f"Common elements between list1 and list2: {common}")
```
