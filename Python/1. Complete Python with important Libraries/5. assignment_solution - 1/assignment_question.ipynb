**Question 19:** Write a Python program to check if a string is a palindrome.

```python
a = input("Enter a string:")
reversed_string = a[::-1]
if a == reversed_string:
    print(f"{a} is an palindrome")
else:
    print(f"{a} is not an palindrome")
```
