and computer programs. OOP allows for modeling real-world scenarios using classes and objects. This lesson
covers the basics of creating classes and objects, including instance variables and methods.

```python
### Instance Variable and Methods
class Dog:
    ## Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

## create objects
dog1 = Dog("Tommy", 2)
print(dog1)
print(dog1.name)
print(dog1.age)
```

```python
## Define a class with instance methods

class Dog:
    # Constructor
    def __init__(self, name, age):
        self.name = name 
        self.age = age

    # Instance method    
    def bark(self):
        print(f"{self.name} says woof!")        

# Create objects            
dog1 = Dog("Tommy", 2)
dog1.bark()
dog2 = Dog("Lucy", 4)
dog2.bark()
```

```python
## Define a class with instance methods (Conditional Statements)

class Dog:
    def __init__(self, name, age, gender):
        self.name = name 
        self.age = age
        self.gender = gender

    def bark(self):
        if self.gender == "Male":
            print(f"{self.name}! He is {self.age} years old and says woof!")
        else:
            print(f"{self.name}! She is {self.age} years old and says woof!")

dog1 = Dog("Tommy", 2, "Male")
dog1.bark()
dog2 = Dog("Lucy", 4, "Female")
dog2.bark()
```

```python
## Modelinf a bank accoiunt

## Define a class for bank account

class BankAccount:
    def __init__(self, owner, balance = 0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print ("Insufficient funds")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New balance is {self.balance}.")

    def get_balance(self):
        return self.balance

## create an account

account1 = BankAccount("Snehashis", 5000)
print(account1.balance)    
```

```python
## Call instance methods
account1.deposit(1000)
```

```python
account1.withdraw(2000)
```

```python
print(account1.get_balance())
```
